# 00. Selección del Universo de Inversión

## Objetivo

Antes de aplicar cualquier técnica de optimización de portafolios, es necesario definir el **universo de inversión**, es decir, el conjunto de activos sobre los cuales trabajará el modelo.

Es importante destacar que los algoritmos de optimización, como la Teoría Moderna de Portafolios (Markowitz), **no seleccionan qué activos analizar**; únicamente determinan la mejor combinación posible de los activos que el analista les proporciona.

Por esta razón, la construcción del universo de inversión constituye una etapa independiente y fundamental dentro del proceso de análisis.

---

## Estrategia de Selección

Para este proyecto, el universo de inversión fue construido mediante una **selección deliberada de activos representativos** de distintos sectores económicos y clases de activos.

El objetivo fue disponer de un conjunto suficientemente diversificado que permitiera demostrar el funcionamiento de las técnicas de optimización y análisis de riesgo desarrolladas posteriormente.

Los principales criterios utilizados fueron:

- Diversificación sectorial.
- Diversificación por clase de activo (acciones, commodities y bonos).
- Inclusión de empresas ampliamente conocidas.
- Disponibilidad de información histórica suficiente.
- Representación de distintos motores económicos.

---

## Alcance del Proyecto

Es importante aclarar que la selección inicial de activos **no forma parte del análisis cuantitativo desarrollado en este proyecto**.

La inclusión de cada empresa o activo **no responde a un análisis fundamental, financiero, contable o macroeconómico específico**, sino a una selección realizada con fines académicos y demostrativos para construir un universo de inversión suficientemente amplio y diversificado.

En consecuencia, este proyecto **no debe interpretarse como una recomendación de inversión sobre los activos seleccionados**, sino como una demostración del proceso analítico utilizado para optimizar y evaluar un portafolio una vez definido el universo de inversión.

---

## Limitaciones

Debido a que el universo fue construido mediante una selección manual de activos, los resultados obtenidos dependen directamente de dicha selección.

La optimización identifica el mejor portafolio posible **dentro del universo disponible**, pero no garantiza que dicho universo represente necesariamente las mejores oportunidades del mercado.

En un entorno profesional, esta etapa suele apoyarse en procesos adicionales como:

- Análisis fundamental.
- Valoración financiera.
- Modelos multifactoriales.
- Indicadores macroeconómicos.
- Filtros de liquidez.
- Restricciones propias del cliente o del mandato de inversión.

---

## Aprendizaje Principal

Uno de los principales aprendizajes de este proyecto es que **la calidad de un portafolio optimizado depende tanto del algoritmo utilizado como de la calidad del universo de inversión seleccionado**.

Ningún modelo matemático puede compensar una selección inicial de activos que no responda a criterios financieros sólidos. La optimización constituye una herramienta de apoyo para la toma de decisiones, pero siempre debe partir de un universo de inversión cuidadosamente definido.

# 00. Investment Universe Selection

## Objective

Before applying any portfolio optimization technique, it is necessary to define the **investment universe**, that is, the set of assets from which the portfolio will be constructed.

It is important to understand that optimization algorithms such as **Modern Portfolio Theory (Markowitz)** do not decide which assets should be analyzed. Instead, they identify the optimal allocation among the assets provided by the analyst.

For this reason, defining the investment universe represents an independent and essential stage of the portfolio construction process.

---

## Selection Strategy

For this project, the investment universe was built through a **judgment-based selection of representative assets** from different economic sectors and asset classes.

The objective was to create a sufficiently diversified universe capable of illustrating portfolio optimization and risk analysis techniques.

The main selection criteria included:

- Sector diversification.
- Asset class diversification (equities, commodities, and bonds).
- Well-known publicly traded companies.
- Availability of sufficient historical market data.
- Exposure to different economic drivers.

---

## Project Scope

It is important to clarify that the initial asset selection **is not part of the quantitative analysis performed in this project**.

The inclusion of each company or financial instrument **was not based on fundamental analysis, financial valuation, accounting analysis, or macroeconomic research**. Instead, the universe was manually constructed for educational and demonstration purposes in order to provide a diversified set of assets for portfolio optimization.

Therefore, this project **should not be interpreted as an investment recommendation regarding the selected assets**, but rather as a demonstration of the analytical workflow used to optimize and evaluate a portfolio once an investment universe has been defined.

---

## Limitations

Because the investment universe was manually selected, the optimization results are inherently dependent on that initial selection.

Portfolio optimization identifies the best allocation **within the available universe**, but it cannot guarantee that the selected universe itself represents the best investment opportunities in the market.

In professional portfolio management, this stage is typically supported by additional analyses such as:

- Fundamental analysis.
- Financial valuation.
- Multi-factor models.
- Macroeconomic analysis.
- Liquidity screening.
- Client-specific investment constraints.

---

## Key Takeaway

One of the main lessons from this project is that **the quality of an optimized portfolio depends not only on the optimization algorithm, but also on the quality of the investment universe provided to that algorithm**.

No mathematical optimization model can compensate for a poorly defined investment universe. Portfolio optimization should therefore be viewed as a decision-support tool built upon a carefully selected set of investment opportunities.
